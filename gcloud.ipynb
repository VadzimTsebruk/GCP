## Troubleshooting

ERROR: There was a problem with web authentication. Try running again with --no-browser.

ERROR: (gcloud.auth.application-default.login) Could not reach the login server.

A potential cause of this could be because you are behind a proxy. Please set the environment variables HTTPS_PROXY and HTTP_PROXY to the address of the proxy in the format "protocol://address:port" (without quotes) and try again.
Example: HTTPS_PROXY=http://192.168.0.1:8080
